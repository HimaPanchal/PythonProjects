```python
#Importing Required libraries
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer
```

```python
my_df = pd.DataFrame({"A": [1,2,3,4,5],
                       "B": [1,1,3,3,4],
                       "C" :[1,2,9,np.nan,20]})
```

```python
knn_imputer = KNNImputer()
knn_imputer.fit_transform(my_df)
```

```python
knn_imputer = KNNImputer(n_neighbors = 1) # Imputing values from the nearest neighbors
knn_imputer.fit_transform(my_df)
```

```python
knn_imputer =KNNImputer(n_neighbors =2) #Imputing values from the nearest 2 neighbors
knn_imputer.fit_transform(my_df)
```

```python
knn_imputer = KNNImputer(n_neighbors = 2, weights ="distance")
knn_imputer.fit_transform(my_df)
```

```python
my_df1 = pd.DataFrame(knn_imputer.fit_transform(my_df),columns=my_df.columns) # Instead of array, converting values in Dataframe
my_df1
```
