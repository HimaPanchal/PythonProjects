```python
#Import Libraries 
import pandas as pd
import numpy as np
```

```python
my_df = pd.DataFrame({"A" : [1,2,4,np.nan,5,np.nan,7],
                      "B" : [4,np.nan,7,np.nan,1,np.nan,2]})
```

# Finding Missing values with Pandas

```python
my_df.isna()
my_df.isna().sum()
```

# Dropping Missing Values with Pandas

```python
my_df.dropna()
my_df.dropna(how="any") # any of the value is missing will drop it
my_df.dropna(how="all") # if all of the value in a row is missing
```

```python
my_df.dropna(how="any",subset=["A"])
```

```python
my_df.dropna(how="any",inplace= True) # To populate full data we use inplace
```

Best Practice is to drop missing values to keep the data being not manipulated but if the dataset is small it is also best practise to fill missing values.

# Filling Missing Values with Pandas

```python
my_df = pd.DataFrame({"A" : [1,2,4,np.nan,5,np.nan,7],
                      "B" : [4,np.nan,7,np.nan,1,np.nan,2]})
```

```python
my_df.fillna(value=10000) #Generally, filling missing values by constant just as 1 or 0 or bigger number to differentiate
```

```python
mean_value = my_df["A"].mean() # Also, best practise is to fill the missing values through mean
```

```python
my_df["A"].fillna(value=mean_value)
```

```python
my_df.fillna(value=my_df.mean(),inplace=True)
```
