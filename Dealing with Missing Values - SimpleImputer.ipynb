```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
```

```python
my_df = pd.DataFrame({"A" : [1,4,7,10,13],
                      "B" : [3,6,9,np.nan,15],
                      "C" : [2,5,np.nan,11,np.nan]})
```

```python
imputer = SimpleImputer()
```

```python
imputer.fit(my_df)
```

```python
imputer.transform(my_df)
```

```python
my_df1 = imputer.transform(my_df)
my_df1
```

```python
#It will also work by adding it together fit and transform
imputer.fit_transform(my_df)
```

```python
#Use this method for only test data 
```

```python
#If you want to work with the Data Frame instead of arrays
my_df2 = pd.DataFrame(imputer.fit_transform(my_df),columns = my_df.columns)
my_df2
```

```python
#Above we applied the imputation logic to all the columns, we may want to apply it to a selection of columns

imputer.fit_transform(my_df[["B"]]) #This will just follow instruction and not overwrite the result dataset
my_df["B"]= imputer.fit_transform(my_df[["B"]]) # This will ensure it overwrites overall result data
my_df
```
